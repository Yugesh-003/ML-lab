```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Temperature (input)
X = np.array([20, 25, 30, 35]).reshape(-1, 1)

# Money Earned (output)
Y = np.array([50, 70, 90, 110])
```

```python
model = LinearRegression()  # Create model
model.fit(X, Y)  # Train model using data
```

```python
temp = np.array([[28]])  # Temperature to predict
predicted_money = model.predict(temp)
print(f"Predicted money earned at 28°C: ${predicted_money[0]:.2f}")
```

```python
plt.scatter(X, Y, color='blue', label="Actual Data")  # Scatter plot of actual data
plt.plot(X, model.predict(X), color='red', label="Prediction Line")  # Regression line
plt.xlabel("Temperature (°C)")
plt.ylabel("Money Earned ($)")
plt.legend()
plt.show()
```
