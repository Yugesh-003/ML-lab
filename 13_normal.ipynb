```python
##### Methods to find the optimal parameters or weights of a line equation   #####
#### The paramters slope and y-intercept should be optimal to get the best fit line. ####
#### There are 2 methods available    #####
#### 1. Normal equation -  Direct method to find the optimal parameters - Useful for small size dataset.#
#### Because it involves matrix multiplication. Hence it is not advisable for large data set.
#### 2. Gradient Descent  -  An Iterative method to find the optimal parameters - Useful for large size dataset.

####  Finding out optimal parameters using normal equation    ####

import numpy as np
import statsmodels.api as s
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

df = pd.read_csv("datasets/Advertising.csv")

x = df['TV']  # Define the independent variable array
y = df['Sales']  # Define the dependent variable array
'''
print(x.shape)
print(x)
'''
ones_column = np.ones((x.shape[0], 1), dtype=int)
#print(ones_column)
x_b = np.c_[ones_column,x] # add x0 = 1 for each instance
# to concatenate arrays along the second axis (columns)
#print(x_b)
#x_b = np.c_[np.ones((34,1)),x] # add x0 = 1 for each instance
print(x_b.shape)

#print(x_b)
theta_best = np.linalg.inv(x_b.T.dot(x_b)).dot(x_b.T).dot(y) # Normal equation
print(theta_best)
```
