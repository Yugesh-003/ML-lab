```python
import numpy as np

# Step 1: Generate 10,000 random data points (e.g., normal distribution)
data = np.random.normal(loc=50, scale=10, size=100000000)  # mean=50, std=10

# Step 2: Compute population variance (divide by N)
population_variance = np.var(data,ddof=1)

print("Population Variance:", population_variance)
```

```python
import numpy as np

# Two arrays
data1 = np.array([10, 15, 20, 25, 30])
data2 = np.array([100, 120, 130, 150, 170])

# Function to compute CV
def coefficient_of_variation(data):
    mean = np.mean(data)
    std = np.std(data)
    return (std / mean) * 100

# Calculate CVs
cv1 = coefficient_of_variation(data1)
cv2 = coefficient_of_variation(data2)

# Output results
print(f"CV of data1: {cv1:.2f}%")
print(f"CV of data2: {cv2:.2f}%")

# Comparison
if cv1 < cv2:
    print("Data1 is more consistent (less relative variation).")
elif cv1 > cv2:
    print("Data2 is more consistent (less relative variation).")
else:
    print("Both datasets have the same relative variation.")
```

```python
import numpy as np

# Market Cap and Earnings
market_cap = np.array([1532, 1488, 1343, 928, 615])
earnings = np.array([58, 35, 75, 41, 17])

# Calculate population covariance
cov_matrix = np.cov(market_cap, earnings, bias=True)  # bias=True for population covariance
cov = cov_matrix[0, 1]

print("Covariance between Market Cap and Earning:", cov)
```
